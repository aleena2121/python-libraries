```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
x = [0,1,2,3,4]
y = [0,2,4,6,8]
# plt.plot(x, y, label="2x", color="red",linewidth = 1, marker=".")
plt.plot(x,y, 'r.--' ,label="2x")
x2 = np.arange(0,4.5,0.5)
plt.plot(x2,x2**2, label="x^2")

plt.title("First Graph",fontdict={'fontname': "Comic Sans MS",'fontsize': 10})
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.xticks([0,1,2,3,4])
plt.yticks([0,2,4,6,8,10])
plt.legend()

plt.savefig('mygraph.png',dpi=300)
plt.show()
```

```python
labels = ["A","B","C"]
values = [1,4,2]
bars = plt.bar(labels,values)
patterns = ["/",".","O"]
for bar in bars:
    bar.set_hatch(patterns.pop())
plt.figure(figsize=(6,4))
plt.show()
```

```python
fifa = pd.read_csv("fifa_data.csv")
fifa.head()
```

```python
bins = [40,50,60,70,80,90,100]
plt.hist(fifa.Overall, bins=bins)
plt.xticks(bins)
plt.ylabel("Number of Players")
plt.xlabel("Skill Level")
plt.title("Distribution of Player skills in FIFA 2018")
plt.show()
```

```python
left = fifa.loc[fifa['Preferred Foot']=='Left'].count()[0]
right = fifa.loc[fifa['Preferred Foot']=='Right'].count()[0]
```

```python
labels=['left','right']
plt.pie([left,right], labels=labels, autopct="%.2f %%")
plt.title("Foot Preference of FIFA Players")
plt.show()
```

```python
fifa.Weight = [int(x.strip('lbs') )if type(x)==str else x for x in fifa.Weight]
```

```python
fifa.Weight
```

```python
light = fifa.loc[fifa.Weight < 125].count()[0]
light_medium = fifa.loc[(fifa.Weight > 125) & (fifa.Weight < 150)].count()[0]
medium = fifa.loc[(fifa.Weight > 150) & (fifa.Weight < 175)].count()[0]
medium_heavy = fifa.loc[(fifa.Weight > 175) & (fifa.Weight < 200)].count()[0]
heavy = fifa.loc[fifa.Weight > 200].count()[0]
```

```python
weights = [light, light_medium, medium, medium_heavy, heavy]
plt.style.use("ggplot")
labels = ["<120","120-150","150-175","175-200",">200"]
explode = (.4,.2,0,0,.4)
plt.pie(weights, labels=labels, autopct="%.2f%%", pctdistance=0.9,explode=explode)    
plt.show()
```

```python
barcelona = fifa.loc[fifa.Club == 'FC Barcelona']['Overall']
madrid = fifa.loc[fifa.Club == 'Real Madrid']['Overall']
revs = fifa.loc[fifa.Club == 'New England Revolution']['Overall']
plt.figure(figsize=(5,8))
plt.style.use("default")
labels = ["FC Barcelona", "Real Madrid", "NE Revolution"]
plt.boxplot([barcelona, madrid, revs], labels=labels)
plt.title("Profesional Soccer Team Comparison")
plt.ylabel("Overall Rating")
plt.show()
```
