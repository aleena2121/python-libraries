14. Write a Python program to display the grid and draw line charts of the closing value of Alphabet Inc. between October 3, 2016 to October 7, 2016. Customized the grid lines with rendering with a larger grid (major grid) and a smaller grid (minor grid).Turn on the grid but turn
off ticks.


```python
plt.title("Financial Data of Alphabet Inc.")
plt.plot(fin_data.Date, fin_data["Close"], "r.-", label="Close")
plt.legend()
plt.minorticks_on()
plt.grid(which="major",linewidth=0.5, linestyle="-",color="blue")
plt.grid(which="minor",linewidth=0.2, linestyle="-",color="blue")
plt.tick_params(axis='both', which='both',
                bottom=False, top=False,
                left=False, right=False,
                labelbottom=False, labelleft=False)
plt.xlabel("Dates")
plt.ylabel("Price")
plt.show()
```

5. Write a Python programming to display a bar chart of the popularity of programming Languages. Attach a text label above each bar displaying its popularity (float value).


```python
languages = ["Java", "Python", "PHP", "JavaScript", "C#", "C++"]
popularity = [22.2, 17.6, 8.8, 8, 7.7, 6.7]

color = ["green", "orange", "lightblue", "yellow", "cyan", "pink"]
plt.bar(languages, popularity, color=color)
for i in range(len(languages)):
    plt.text(i,popularity[i]+0.2,popularity[i])
plt.xlabel("Languages")
plt.ylabel("Popularity")
plt.title("Popularity of Languages")
plt.show()
```

10. Write a Python program to create bar plot of scores by group and gender. Use multiple X values on the same chart for men and women.


```python
men = [22, 30, 35, 35, 26]
women = [25, 32, 30, 35, 29]
x = np.arange(5)
bar_width = 0.35
plt.bar(x-bar_width/2,men, width=bar_width, label="Men")
plt.bar(x+bar_width/2,women, width=bar_width, label="Women")
plt.xlabel("Groups")
plt.ylabel("Means")
plt.legend()
plt.title("Mean Scores of Men and Women")
plt.show()
```

12. Write a Python program to create bar plots with error bars on the same figure.

```python
mean = [0.2474, 0.1235, 0.1737, 0.1824]
sd = [0.3314, 0.2278, 0.2836, 0.2645]
x = np.arange(4)
plt.bar(x,mean, yerr=sd, capsize=5)
plt.xticks(x)
plt.xlabel("Groups")
plt.ylabel("Means")
plt.title("Means and Standard Deviation")
plt.show()
```

13. Write a Python program to create bar plots with errorbars on the same figure. Attach a text label above each bar displaying men means.


```python
mean = [0.2474, 0.1235, 0.1737, 0.1824]
sd = [0.3314, 0.2278, 0.2836, 0.2645]
x = np.arange(4)
plt.bar(x,mean, yerr=sd, capsize=5)
plt.xticks(x)
for i in range(len(x)):
    plt.text(i-0.2,mean[i]+0.03,mean[i])
plt.xlabel("Groups")
plt.ylabel("Means")
plt.title("Means and Standard Deviation")
plt.show()
```
