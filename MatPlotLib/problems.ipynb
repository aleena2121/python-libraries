3. Write a Python program to draw a line using given axis values taken from a text file, with suitable label in the x axis, y axis and a title.


```python
x = []
y = []
with open("Files/test.txt","r") as file:
    nums = file.readlines()
    for num in nums:
        x.append(num[0])
        y.append(num[2])
plt.title("Plotting values from file")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.plot(x,y)
plt.show()
```

12. Write a Python program to create multiple types of charts

```python
fig, axs = plt.subplots(2,2, figsize=(10,8))
axs[0][0].plot(fin_data.Date, fin_data["Open"], "b.--", label="Open")
axs[0][0].set_title("Open")

axs[0][1].plot(fin_data.Date, fin_data["High"], "r.:", label="High")
axs[0][1].set_title("High")

axs[1][0].plot(fin_data.Date, fin_data["Low"], "y.-.", label="Low")
axs[1][0].set_title("Low")

axs[1][1].plot(fin_data.Date, fin_data["Close"], "g.-", label="Close")
axs[1][1].set_title("Close")

plt.tight_layout()
plt.show()
```

13. Write a Python program to display the grid and draw line charts of the closing value of Alphabet Inc. between October 3, 2016 to October 7, 2016. Customized the grid lines with linestyle -, width .5. and color blue.

14. Write a Python program to display the grid and draw line charts of the closing value of Alphabet Inc. between October 3, 2016 to October 7, 2016. Customized the grid lines with rendering with a larger grid (major grid) and a smaller grid (minor grid).Turn on the grid but turn
off ticks.


```python
plt.title("Financial Data of Alphabet Inc.")
plt.plot(fin_data.Date, fin_data["Close"], "r.-", label="Close")
plt.legend()
plt.minorticks_on()
plt.grid(which="major",linewidth=0.5, linestyle="-",color="blue")
plt.grid(which="minor",linewidth=0.2, linestyle="-",color="blue")
plt.tick_params(axis='both', which='both',
                bottom=False, top=False,
                left=False, right=False,
                labelbottom=False, labelleft=False)
plt.xlabel("Dates")
plt.ylabel("Price")
plt.show()
```

15. Write a Python program to create multiple plots.

```python
x = [1,2,3,4,5,6]
y = [2,4,6,7,10,12]
z = [6,9,10,15,30,33]

plt.figure(figsize=(8,8))

plt.subplot(2,2,1)
plt.plot(x,y, marker=".")

plt.subplot(2,2,2)
plt.plot(y, x, marker=".")

plt.subplot(2,2,3)
plt.plot(y, z, marker=".")

plt.subplot(2,2,4)
plt.plot(z, z, marker=".")

plt.show()
```
