7. Write a Python program to create a 8x8 matrix and fill it with a checkerboard pattern.

```python
checkerboard = np.zeros([8,8], dtype=int)
for i in range(8):
    for j in range(8):
        if (i+j) % 2 == 1:
            checkerboard[i][j] = 1
checkerboard
```

11. Write a Python program to find the number of elements of an array, length of one array element in bytes and total bytes consumed by the elements.


```python
arr = np.array([1,4,6])
print(f"Size of the array : {arr.size}")
print(f"Length of one array element in bytes : {arr.itemsize}")
print(f"Total bytes consumed by the elements of the array : {arr.itemsize * arr.size}")
```

12. Write a Python program to find common values between two arrays

```python
arr1 = np.array([0, 10, 20, 40, 60])
arr2 = np.array([10, 30, 40])
common = np.intersect1d(arr1,arr2)
print(f"Common Elements : {common}")
```

13. Write a Python program to find the set difference of two arrays

```python
arr1 = np.array([0, 10, 20, 40, 60, 80])
arr2 = np.array([10, 30, 40, 50, 70, 90])
common = np.setdiff1d(arr1,arr2)  
print(f"Set Difference : {common}")
```

14. Write a Python program to find the set exclusive-or of two arrays

```python
arr1 = np.array([0, 10, 20, 40, 60, 80])
arr2 = np.array([10, 30, 40, 50, 70, 90])
xor = np.setxor1d(arr1,arr2)  
print(f"Set Exclusive : {xor}")
```

15. Write a Python program compare two arrays using numpy.

```python
a = np.array([1,3])
b = np.array([4,5])

print(f"a > b : {np.greater(a,b)}")
print(f"a >= b : {np.greater_equal(a,b)}")
print(f"a < b : {np.less(a,b)}")
print(f"a <= b : {np.less_equal(a,b)}")
```
