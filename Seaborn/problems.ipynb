```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Write a program to draw bar plot of sex against survived

```python
titanic_df = sns.load_dataset('titanic')
titanic_df
```

```python
sns.barplot(x='sex', y='survived', data=titanic_df)
```

Write a program to draw a point plot for sex against survived

```python
sns.pointplot(x='sex', y='survived', data=titanic_df)
```

Write a program to draw a scatter plot of “day” against “total bill”

```python
tips_df = sns.load_dataset('tips')
tips_df.head()
```

```python
sns.scatterplot(x="day", y="total_bill", data=tips_df)
```

Write a program to draw a violin plot of sex against total_bill


```python
sns.violinplot(x='sex', y='total_bill', data=tips_df)
```
